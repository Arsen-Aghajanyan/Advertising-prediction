```python
import numpy as np
import pandas as pd
import seaborn as sns
import plotly.express as px
```

```python
ds = pd.read_csv('advertising.csv')
```

```python
ds.head()
```

```python
fig = px.scatter(data_frame=ds, x='TV', y='Sales', title='TV Sales')
fig.show()
```

```python
fig = px.scatter(data_frame=ds, x="Radio", y="Sales", title='Radio Sales')
fig.show()
```

```python
fig = px.scatter(data_frame=ds, x='Newspaper', y='Sales', title='Newspaper Sales')
fig.show()
```

```python
fig = px.pie(
    names=['TV', 'Radio', 'Newspaper'],
    values=[ds['TV'].sum(), ds['Radio'].sum(), ds['Newspaper'].sum()],
    title='Total budget'
)
fig.show()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = ds.drop('Sales', axis=1)
y = ds['Sales']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
accuracy = model.score(X_test, y_test)
print("Accuracy (R^2):", accuracy)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
example = pd.DataFrame({
    "TV": [150],
    "Radio": [20],
    "Newspaper": [30],
})
pred = model.predict(example)
print("Predicted sales:", pred[0])
```

```python
fig = px.scatter(x=y_test, y=y_pred, labels={
    'X': "Actual sales",
    'Y': "Predicted sales",
}, title='Actual & Pred sales')

fig.add_shape(
    type="line",
    x0 = y_test.min(), y0 = y_test.min(),
    x1 = y_test.max(), y1 = y_test.max(),
)
fig.show()
```
